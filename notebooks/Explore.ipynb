## Retrieving Time Series Data

Once your `FulcraAPI` instance is authorized, you can now use it
to make API calls.

The `metric_time_series` function returns a pandas DataFrame for a single metric, indexed by time. To work with several metrics at once, fetch each one and join them with `pd.concat`.

### Available Metrics
Call `fulcra.metrics_catalog()` for the full list of available metrics and their descriptions.

```python
# Try changing the start/end times
start = '2023-09-01 04:00:00.000Z'
end = '2023-09-15 04:00:00.000Z'

# Metrics to retrieve
metric_names = [
    'StepCount',
    'HeartRate',
    'ActiveCaloriesBurned',
    'AppleWatchExerciseTime'
]

# metric_time_series returns one metric at a time; fetch each one and join them
# into a single dataframe on their shared time index.
# Note: omit sample_rate to use the default of one sample per minute
df = pd.concat(
    [
        fulcra.metric_time_series(start_time=start, end_time=end, metric=metric)
        for metric in metric_names
    ],
    axis=1,
)

# inspect returned data frame
df
```

## Optional Step
Add columns to the dataframe so we have more date-related information to use when creating a chart.

**Note**: *this is optional because it can also be achieved via PyGWalker by clicking the triple-dot icon for the `date` field in **Field List** after running the next cell:*

```python
df['year'] = df.index.year
df['month'] = df.index.month
df['day'] = df.index.day
df['weekdayName'] = df.index.day_name()
df['weekday'] = df.index.weekday
df['week'] = df.index.isocalendar().week
df['weekend'] = df.index.weekday // 5 == 1
df['date'] = df.index
df['hour'] = df.index.hour
```

This demo uses `PyGWalker`: https://github.com/Kanaries/pygwalker

Use the PyGWalker `walk` method to create a graphing interface:

```python
pyg.walk(df)
```
